Design Inquiry

The Bayes optimal classifier combines predictions from multiple hypotheses. The majority-weighted decision may therefore differ from the prediction of every individual hypothesis.

Why is it rarely used for large hypothesis spaces?

It may require:

Maintaining many hypotheses.

Computing their posterior probabilities.

Combining their predictions.

Performing expensive inference.

What is the advantage of Gibbs?

Gibbs samples one hypothesis rather than combining all hypotheses. This can make prediction easier when posterior sampling is available.

return (

1 / np.sqrt(2 * np.pi * var)

) * np.exp(

-(x - mu)**2 / (2 * var)

)

def em_gmm(X, K=2, seed=42,

max_iter=500, tol=1e-6):

n = len(X)

pi = np.ones(K) / K

mu = rng.choice(X, K, replace=False).astype(float)

var = np.full(K, np.var(X))

# E-step

weighted = np.column_stack([

pi[k] * gaussian_pdf(X, mu[k], var[k])

for k in range(K)

])

weighted /

weighted.sum(axis=1, keepdims=True)

)

responsibilities.T @ X

) / Nk

responsibilities.T @ (

(X[:, None] - mu)**2

)

pi[k] * gaussian_pdf(X, mu[k], var[k])

for k in range(K)

])

ll = np.log(weighted.sum(axis=1)).sum()

history.append(ll)

if len(history) > 1:

if abs(history[-1] - history[-2]) < tol:

break

return pi, mu, var, history

6.13 PAC Learning Versus Mistake-Bound Learning

| Aspect | PAC Learning | Mistake-Bound Learning |
| --- | --- | --- |
| Main objective | Generalisation accuracy | Limit prediction mistakes |
| Data | Usually i.i.d. examples | Sequential examples |
| Guarantee | Error probability | Maximum mistakes |
| Parameters | ε and δ | Hypothesis class size |
| Main quantity | Sample complexity | Mistake complexity |
| Application | Batch learning | Online learning |

Which guarantee is more useful for the quality team?

For a quality team screening bearings, PAC learning is useful when the objective is to determine how many labelled inspection records are needed to achieve a specified error rate with a specified confidence.

For a continuously operating production line, the mistake-bound model is useful when examples arrive sequentially and every incorrect prediction has an operational cost.

Therefore:

PAC is more useful for planning inspection-data collection.

Mistake bounds are more useful for analysing online learning performance.

10. View and Copy in NumPy

A view creates a different way of accessing the same underlying data, while a copy creates independent data.

Example:

original = np.array([10, 20, 30, 40, 50])

view_array = original[1:4]

copy_array = original[1:4].copy()

view_array[0] = 100

copy_array[1] = 200

print("Original:", original)

print("View:", view_array)

print("Copy:", copy_array)

Output:

Original: [ 10 100  30  40  50]

View: [100  30  40]

Copy: [ 30 200  40]

The view changes the original array because both share the same underlying data. The copy does not affect the original array.

5. Identify Unusual Readings Using Z-Score

Z-score identifies values that are far from the mean.

\[ Z=\frac{x-\mu}{\sigma} \]

A reading is considered unusual when:

\[ |Z|>3 \]

mean = np.mean(data)

std = np.std(data)

z_scores = (data - mean) / std

outlier_mask = np.abs(z_scores) > 3

rows, columns = np.where(outlier_mask)

print("Outlier Rows:", rows)

print("Outlier Columns:", columns)

print("Outlier Values:", data[rows, columns])

The expected inserted extreme positions are:

(2, 5)

(10, 15)

(20, 25)

(35, 8)

(45, 20)

The actual detected positions are determined by the z-score calculation.

7. Compare Loop-Based and Vectorised Calculations

)

# Vectorised method

Explanation: Vectorised calculations are generally faster because NumPy performs operations using optimised internal routines instead of repeatedly executing Python-level loops.

The original inner merge removes unmatched records, whereas the revised left merge preserves them.

3. New Costs

The redesigned method requires additional validation and storage for unmatched records.

However, the additional cost is reasonable because preserving production information is more important than reducing a small amount of processing time.

4. Final Design Decision

The validated left merge is selected as the final strategy because it preserves production records and identifies missing machine information.

The inner merge may still be preferable when an analysis specifically requires records with complete matching information from both datasets.

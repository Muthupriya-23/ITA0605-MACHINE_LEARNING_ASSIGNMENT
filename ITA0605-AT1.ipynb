TASK 7: HEURISTIC SEARCH
ID3 starts with an empty tree/root containing all rows. An operator selects an unused attribute and
partitions the current subset. The heuristic is information gain; termination occurs at a pure node, no
available attributes, or the chosen stopping threshold. At a node, candidate computations equal the number
of unused attributes. Total computations are the sum across decision nodes.

Gain ratio = information gain / split information, where split information = −Σ p(v)log2 p(v). Gain ratio
penalises attributes with many tiny partitions. A unique StudentID creates 14 singleton pure leaves, giving
training information gain 1.0000, but it memorises rows and cannot generalise.

An Attendance-rooted alternative is the greedy tree if Attendance is selected at root; compare it with the
chosen tree using node/leaf count, depth and accuracy from the deterministic implementation. ID3 is
greedy and does not backtrack: it may miss a globally smaller tree. Advantages: fast, interpretable
partitions. Limitations: greedy local optimum and high variance on small/noisy data.
